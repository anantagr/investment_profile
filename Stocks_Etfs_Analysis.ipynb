```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import HTML, display
import requests
from bs4 import BeautifulSoup
import time
from requests.exceptions import RequestException
from typing import Dict, List, Optional

# Define Stock and ETF Tickers
stock_symbols = ["RY.TO", "TD.TO", "BMO.TO", "BNS.TO", "CM.TO", "SHOP.TO", "ENB.TO", "CNQ.TO", "CVE.TO", "MFC.TO", "WSHR.NE"]
etf_symbols = ["XIC.TO", "XIU.TO", "XBB.TO", "XEQT.TO", "ZSP.TO"]

USER_AGENT = "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36"

def fetch_data(ticker: str) -> Optional[Dict[str, str]]:
    """
    Fetches stock or ETF data for a given ticker from Yahoo Finance.

    Args:
        ticker: The stock or ETF ticker symbol (e.g., "RY.TO").

    Returns:
        A dictionary containing financial data for the ticker, or None if data fetching fails.
        Keys include "Company", "Ticker", "Current Price", "Open", "Last Sale", etc.
    """
    try:
        asset = yf.Ticker(ticker)
        info = asset.info

        current_price = info.get('regularMarketPrice', info.get('previousClose'))
        open_price = info.get('open')
        last_sale = info.get("previousClose")

        fifty_two_week_high = info.get("fiftyTwoWeekHigh")
        fifty_two_week_low = info.get("fiftyTwoWeekLow")
        day_high = info.get('dayHigh')
        day_low = info.get('dayLow')

        volume = info.get("volume")
        avg_volume = info.get("averageVolume")
        market_cap = info.get("marketCap")
        pe_ratio = info.get("trailingPE")
        forward_pe = info.get("forwardPE")
        eps = info.get("trailingEps")
        expense_ratio = info.get("expenseRatio")
        yield_percent = info.get("yield")
        bid = info.get("bid")
        ask = info.get("ask")
        exchange = info.get("exchange")

        change_percent = ((current_price - open_price) / open_price) * 100 if open_price else None
        change_dollar = current_price - open_price if open_price else None

        percent_from_high = ((last_sale - fifty_two_week_low) / (fifty_two_week_high - fifty_two_week_low)) * 100 if fifty_two_week_high and fifty_two_week_low else None

        return {
            "Company": info.get("shortName", ticker),
            "Ticker": ticker,
            "Current Price": f"${current_price:.2f}" if current_price else None,
            "Open": f"${open_price:.2f}" if open_price else None,
            "Last Sale": f"${last_sale:.2f}" if last_sale else None,
            "Day Change (%)": f"{change_percent:.2f}%" if change_percent else None,
            "Day Change ($)": f"${change_dollar:.2f}" if change_dollar else None,
            "Day's Range": f"${day_low:.2f} - ${day_high:.2f}" if day_low and day_high else None,
            "52 Week Range": f"${fifty_two_week_low:.2f} - ${fifty_two_week_high:.2f}" if fifty_two_week_low and fifty_two_week_high else None,
            "Bid": f"${bid:.2f}" if bid else None,
            "Ask": f"${ask:.2f}" if ask else None,
            "High": f"${day_high:.2f}" if day_high else None,
            "Low": f"${day_low:.2f}" if day_low else None,
            "Volume": f"{volume:,}" if volume else None,
            "Avg. Volume": f"{avg_volume:,}" if avg_volume else None,
            "Market Cap (B)": f"${round(market_cap / 1e9, 2)}B" if market_cap else None,
            "P/E Ratio": f"{pe_ratio:.2f}" if pe_ratio else "N/A",
            "Forward P/E": f"{forward_pe:.2f}" if forward_pe else "N/A",
            "EPS (TTM)": f"${eps:.2f}" if eps else None,
            "Management Expense Ratio": f"{expense_ratio:.2f}%" if expense_ratio else "N/A",
            "Exchange": exchange,
            "Yield (%)": f"{yield_percent:.2f}%" if yield_percent else "N/A",
            "52W High": f"${fifty_two_week_high:.2f}" if fifty_two_week_high else None,
            "52W Low": f"${fifty_two_week_low:.2f}" if fifty_two_week_low else None,
            "52W % from High": f"{percent_from_high:.2f}%" if percent_from_high else None,
        }
    except Exception as e:
        print(f"Error fetching data for {ticker}: {e}")
        return None

def fetch_all_data(tickers: List[str]) -> List[Optional[Dict[str, str]]]:
    """
    Fetches data for a list of stock or ETF tickers.

    Args:
        tickers: A list of stock or ETF ticker symbols.

    Returns:
        A list of dictionaries, where each dictionary contains financial data for a ticker,
        or None if data fetching failed for a ticker.
    """
    return [fetch_data(ticker) for ticker in tickers]

def color_change(val: str) -> str:
    """
    Applies color formatting to percentage changes in HTML.

    Positive changes are displayed in green, negative changes in red.

    Args:
        val: A string representing a percentage change (e.g., "2.50%").

    Returns:
        An HTML string with a span element that applies color formatting,
        or the original value if it's not a valid percentage change.
    """
    try:
        val_num = float(val.replace('%', '')) # Remove '%' and convert to float
        if val_num > 0:
            return '<span class="positive">{}</span>'.format(val) # Green for positive change
        elif val_num < 0:
            return '<span class="negative">{}</span>'.format(val) # Red for negative change
        else:
            return val # No color for zero change
    except ValueError:
        return val # Return original value if not convertible to float

def display_data_table(df: pd.DataFrame, title: str):
    """
    Displays stock/ETF data in a styled HTML table with color-coded percentage changes.

    Args:
        df: DataFrame containing stock or ETF data.
        title: Title of the table (e.g., "Stock Data", "ETF Data").
    """
    print(f"\n📊 {title}:\n")
    css_style = """
    <style>
    .positive { color: green; }
    .negative { color: red; }
    </style>
    """
    html = df.to_html(
        classes='table-bordered',
        formatters={'Day Change (%)': color_change},
        escape=False
    )
    display(HTML(css_style + html))

def plot_pe_ratio(df: pd.DataFrame, title: str):
    """
    Plots a bar chart comparing the P/E Ratios of selected stocks or ETFs.

    Args:
        df: DataFrame containing stock or ETF data with a 'Company' and 'P/E Ratio' column.
        title: Title of the plot (e.g., "Stocks", "ETFs").
    """
    plt.figure(figsize=(12, 5))
    pe_ratios = pd.to_numeric(df["P/E Ratio"], errors='coerce')
    plt.bar(df["Company"], pe_ratios, color='skyblue')
    plt.xlabel("Company")
    plt.ylabel("P/E Ratio")
    plt.title(f"P/E Ratio Comparison of {title}")
    plt.xticks(rotation=45, ha='right')
    plt.grid(axis="y", linestyle="--", alpha=0.7)
    plt.tight_layout() # Added to ensure labels fit in the plot
    plt.show()

def scrape_holdings_from_html(html_content: bytes) -> Optional[pd.DataFrame]:
    """
    Scrapes top holdings data from HTML content of a Yahoo Finance ETF page.

    Args:
        html_content: HTML content as bytes, fetched from Yahoo Finance ETF holdings page.

    Returns:
        DataFrame: DataFrame with columns ["Symbol", "Company", "% Assets"] containing top holdings data,
                   or None if scraping fails or no holdings data is found.
    """
    soup = BeautifulSoup(html_content, "html.parser")
    holdings_section = soup.find("section", {"data-testid": "top-holdings"})

    if holdings_section:
        data = []
        for content_div in holdings_section.find_all("div", class_="content yf-1ix710n"):
            symbol_span = content_div.find("span", class_="symbol yf-1hqcuho")
            symbol = symbol_span.text.strip() if symbol_span else None

            company_span = content_div.find("span", class_="name yf-1ix710n")
            company = company_span.text.strip() if company_span else None

            holding_span = content_div.find("span", class_="data yf-1ix710n")
            holding_text = holding_span.text.strip() if holding_span else None
            holding_percentage = _clean_and_convert_percentage(holding_text)

            data.append([symbol, company, holding_percentage])

        df = pd.DataFrame(data, columns=["Symbol", "Company", "% Assets"])
        return df
    return None

def scrape_sector_weights(html_content: bytes) -> Optional[pd.DataFrame]:
    """
    Scrapes sector weights data from HTML content of a Yahoo Finance ETF page.

    Args:
        html_content: HTML content as bytes, fetched from Yahoo Finance ETF page.

    Returns:
        DataFrame: DataFrame with columns ["Sector", "Weight"] containing sector weights data,
                   or None if scraping fails or no sector weights data is found.
    """
    soup = BeautifulSoup(html_content, "html.parser")
    sector_weights_section = soup.find("section", {"data-testid": "etf-sector-weightings-overview"})

    if sector_weights_section:
        data = []
        for content_div in sector_weights_section.find_all("div", class_="content yf-szlt12"):
            sector_link = content_div.find("a", class_="primary-link fin-size-medium ellipsis yf-1xqzjha")
            sector = sector_link.text.strip() if sector_link else None

            weight_span = content_div.find("span", class_="data yf-szlt12")
            weight_text = weight_span.text.strip() if weight_span else None
            weight_percentage = _clean_and_convert_percentage(weight_text)

            data.append([sector, weight_percentage])

        df = pd.DataFrame(data, columns=["Sector", "Weight"])
        return df
    return None

def _clean_and_convert_percentage(value: Optional[str]) -> Optional[float]:
    """
    Cleans and converts a percentage string to a float.

    Removes '%' sign and converts to float, handling potential ValueErrors and None inputs.

    Args:
        value: A string representing a percentage (e.g., "2.50%"), or None.

    Returns:
        A float representing the percentage (e.g., 0.0250), or None if conversion fails.
    """
    if value:
        cleaned_value = value.replace("%", "").strip()
        try:
            return float(cleaned_value) / 100
        except ValueError:
            return None
    return None

def fetch_html_with_retry(url: str, retries: int = 3, wait_time: int = 1) -> Optional[bytes]:
    """
    Fetches HTML content from a URL with retry mechanism to handle potential request issues.

    Includes handling for "Too Many Requests" (429) errors with exponential backoff.

    Args:
        url: URL to fetch.
        retries: Maximum number of retry attempts.
        wait_time: Initial wait time in seconds before retrying.

    Returns:
        HTML content as bytes, or None if fetching fails after maximum retries.
    """
    headers = {"User-Agent": USER_AGENT}
    for _ in range(retries):
        try:
            response = requests.get(url, headers=headers)
            response.raise_for_status()  # Raise HTTPError for bad responses (4xx or 5xx)
            return response.content
        except RequestException as e:
            if response and response.status_code == 429:
                print(f"Too many requests. Retrying in {wait_time} seconds...")
                time.sleep(wait_time)
                wait_time *= 2 # Exponential backoff
            else:
                print(f"Error fetching URL {url}: {e}")
                return None
    print(f"Max retries exceeded for URL {url}. Unable to fetch content.")
    return None

def plot_holdings_chart(df: pd.DataFrame, etf_symbol: str):
    """
    Plots a bar chart of top holdings for a given ETF.

    Args:
        df: DataFrame containing ETF holdings data with 'Company' and '% Assets' columns.
        etf_symbol: ETF ticker symbol for chart title.
    """
    if df is None or df.empty:
        print(f"No holdings data to plot for {etf_symbol}.")
        return

    df["% Assets"] = df["% Assets"].astype(str).str.replace("%", "").astype(float)

    plt.figure(figsize=(12, 6))
    bars = plt.bar(df["Company"], df["% Assets"], color='skyblue')

    for bar in bars:
        yval = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2, yval, f"{round(yval, 2)}%", va='bottom', ha='center', fontsize=9)

    plt.xlabel("Company")
    plt.ylabel("% Assets")
    plt.title(f"Top Holdings for {etf_symbol}")
    plt.xticks(rotation=45, ha='right', fontsize = 9)
    plt.tight_layout()
    plt.show()

def plot_sector_weights_chart(df: pd.DataFrame, etf_symbol: str):
    """
    Plots a bar chart of sector weights for a given ETF.

    Args:
        df: DataFrame containing ETF sector weights data with 'Sector' and 'Weight' columns.
        etf_symbol: ETF ticker symbol for chart title.
    """
    if df is None or df.empty:
        print(f"No sector weights data to plot for {etf_symbol}.")
        return

    df["Weight"] = df["Weight"].astype(str).str.replace("%", "").astype(float)

    plt.figure(figsize=(10, 6))
    bars = plt.bar(df["Sector"], df["Weight"], color='lightgreen')

    for bar in bars:
        yval = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2, yval, f"{round(yval, 2)}%", va='bottom', ha='center')

    plt.xlabel("Sector")
    plt.ylabel("Weight")
    plt.title(f"Sector Weights for {etf_symbol}")
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

def scrape_etf_data(etf_symbol: str):
    """
    Scrapes and displays top holdings and sector weights data for a given ETF symbol from Yahoo Finance.

    Args:
        etf_symbol: The ETF ticker symbol (e.g., "XIC.TO").
    """
    url = f"https://finance.yahoo.com/quote/{etf_symbol}/holdings/"
    html_content = fetch_html_with_retry(url)

    if html_content:
        holdings_df = scrape_holdings_from_html(html_content)
        sector_weights_df = scrape_sector_weights(html_content)

        if holdings_df is not None:
            print(f"\nTop Holdings for {etf_symbol}:")
            if holdings_df.empty:
                print("No holdings data found.")
            else:
                holdings_df["% Assets"] = (holdings_df["% Assets"] * 100).round(2).astype(str) + "%" # Correctly format percentage for display
                print(holdings_df)
                plot_holdings_chart(holdings_df.copy(), etf_symbol)

        if sector_weights_df is not None:
            print(f"\nSector Weights for {etf_symbol}:")
            if sector_weights_df.empty:
                print("No sector weights data found.")
            else:
                sector_weights_df["Weight"] = (sector_weights_df["Weight"] * 100).round(2).astype(str) + "%" # Correctly format percentage for display
                print(sector_weights_df)
                plot_sector_weights_chart(sector_weights_df.copy(), etf_symbol)

    else:
        print(f"Failed to get HTML content for {etf_symbol} after retries.")

if __name__ == "__main__":
    # Fetch and display stock data
    stock_data = fetch_all_data(stock_symbols)
    df_stocks = pd.DataFrame(stock_data).dropna(axis=1, how='all') # Drop columns if all values are NaN
    if not df_stocks.empty:
        display_data_table(df_stocks, "Stock Data")
        plot_pe_ratio(df_stocks, "Stocks")
    else:
        print("No stock data fetched successfully to display.")


    # Fetch and display ETF data
    etf_data = fetch_all_data(etf_symbols)
    df_etfs = pd.DataFrame(etf_data).dropna(axis=1, how='all') # Drop columns if all values are NaN
    if not df_etfs.empty:
        display_data_table(df_etfs, "ETF Data")
        plot_pe_ratio(df_etfs, "ETFs")
    else:
        print("No ETF data fetched successfully to display.")

    # Scrape and display ETF holdings and sector weights
    for etf_symbol in etf_symbols:
        scrape_etf_data(etf_symbol)
        print("-" * 50)
```
