```python
# 📌 Install necessary libraries (Run this first in Google Colab)
!pip install yfinance pandas matplotlib --quiet

# ✅ Import Libraries
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import HTML, display
from datetime import date, timedelta

# ✅ Define Stock Tickers to Compare (Modify this list as needed)
stock_symbols = ["RY.TO", "TD.TO", "BMO.TO", "BNS.TO", "CM.TO", "SHOP.TO", "ENB.TO", "CNQ.TO", "CVE.TO", "MFC.TO", "WSHR.NE"]

# ✅ Print Definitions of Stock Metrics
metric_definitions = {
    "Bid": "The highest price a buyer is willing to pay for the stock.",
    "Ask": "The lowest price a seller is willing to accept for the stock.",
    "Last Sale": "The most recent price at which the stock was traded.",
    "Open": "The stock's opening price for the trading session.",
    "High": "The highest price reached during the session.",
    "Low": "The lowest price reached during the session.",
    "52W High": "The highest stock price recorded in the past 52 weeks.",
    "52W Low": "The lowest stock price recorded in the past 52 weeks.",
    "Market Cap (B)": "The total market value of the company's outstanding shares, measured in billions.",
    "P/E Ratio": "The price-to-earnings ratio, which indicates how much investors are willing to pay for $1 of earnings.",
    "Forward P/E": "The forward price-to-earnings ratio, which is calculated using future earnings estimates.",
    "EPS (TTM)": "Earnings per share over the trailing twelve months.",
    "Management Expense Ratio": "The annual fee charged by a fund to manage investors' money.",
    "Net Profit Margin": "A financial metric used to assess a company's profitability.",
    "Volume": "The total number of shares traded during the session.",
    "Avg. Volume": "The average daily trading volume over a period.",
    "Exchange": "The stock exchange where the stock is traded.",
    "Yield (%)": "The dividend income as a percentage of the stock price.",
    "Margin Requirement": "The percentage of the stock price that must be maintained as collateral if purchased on margin."
}

def print_metric_definitions():
    """Prints the definitions of stock metrics."""
    print("\n📚 Stock Metrics Definitions:\n")
    for metric, definition in metric_definitions.items():
        print(f"🔹 {metric}: {definition}")

# ✅ Function to Fetch Stock Data from Yahoo Finance
def fetch_stock_data(ticker):
    """Fetches stock data for a given ticker."""
    stock = yf.Ticker(ticker)
    info = stock.info

    # Get the current price (using 'regularMarketPrice' if available, otherwise 'previousClose')
    current_price = info.get('regularMarketPrice', info.get('previousClose'))
    open_price = info.get('open')
    last_sale = info.get("previousClose")

    fifty_two_week_high = info.get("fiftyTwoWeekHigh")
    fifty_two_week_low = info.get("fiftyTwoWeekLow")
    day_high = info.get('dayHigh')
    day_low = info.get('dayLow')

    volume = info.get("volume")
    formatted_volume = "{:,}".format(volume) if volume is not None else None
    avg_volume = info.get("averageVolume")
    formatted_avg_volume = "{:,}".format(avg_volume) if avg_volume is not None else None

    market_cap = info.get("marketCap")
    pe_ratio = info.get("trailingPE")
    forward_pe = info.get("forwardPE")
    eps = info.get("trailingEps")
    management_expense_ratio = info.get("expenseRatio")

    change_1d_percent = ((current_price - open_price) / open_price) * 100 if (open_price is not None and open_price != 0 and current_price is not None) else None
    change_1d_dollar = current_price - open_price if (open_price is not None and current_price is not None) else None

    days_range = f"${day_low:.2f} - ${day_high:.2f}" if (day_low is not None and day_high is not None) else None
    week_range = f"${fifty_two_week_low:.2f} - ${fifty_two_week_high:.2f}" if (fifty_two_week_low is not None and fifty_two_week_high is not None) else None

    return {
        "Company": info.get("shortName", ticker),
        "Ticker": ticker,
        "Current Price": f"${current_price:.2f}" if current_price is not None else None,
        "Open": f"${open_price:.2f}" if open_price is not None else None,
        "Last Sale": f"${last_sale:.2f}" if last_sale is not None else None,
        "Day Change (%)": f"{change_1d_percent:.2f}%" if change_1d_percent is not None else None,
        "Day Change ($)": f"{change_1d_dollar:.2f}" if change_1d_dollar is not None else None,
        "Day's Range": days_range,
        "52 Week Range": week_range,
        "Bid": f"${info.get('bid'):.2f}" if info.get('bid') is not None else None,
        "Ask": f"${info.get('ask'):.2f}" if info.get('ask') is not None else None,
        "High": f"${info.get('dayHigh'):.2f}" if info.get('dayHigh') is not None else None,
        "Low": f"${info.get('dayLow'):.2f}" if info.get('dayLow') is not None else None,
        "Volume": formatted_volume,
        "Avg. Volume": formatted_avg_volume,
        "Market Cap (B)": f"${round(market_cap / 1e9, 2)}B" if market_cap is not None else None,
        "P/E Ratio": f"{pe_ratio:.2f}" if pe_ratio is not None else "N/A",
        "Forward P/E": f"{forward_pe:.2f}" if forward_pe is not None else "N/A",
        "EPS (TTM)": f"${eps:.2f}" if eps is not None else None,
        "Management Expense Ratio": f"{management_expense_ratio:.2f}%" if management_expense_ratio is not None else "N/A",
        "Exchange": info.get("exchange"),
        "Yield (%)": f"{round(info.get('dividendYield', 0) * 100, 2)}%" if info.get('dividendYield') is not None else None,
        "52W High": f"${fifty_two_week_high:.2f}" if fifty_two_week_high is not None else None,
        "52W Low": f"${fifty_two_week_low:.2f}" if fifty_two_week_low is not None else None,
    }

def fetch_all_stock_data(tickers):
    """Fetches stock data for all tickers."""
    return [fetch_stock_data(ticker) for ticker in tickers]

def calculate_52w_percent_from_high(df):
    """Calculates the percentage from the 52-week high."""
    df["52W % from High"] = ((pd.to_numeric(df["Last Sale"].str.replace('$', ''), errors='coerce') - pd.to_numeric(df["52W Low"].str.replace('$', ''), errors='coerce')) / (pd.to_numeric(df["52W High"].str.replace('$', ''), errors='coerce') - pd.to_numeric(df["52W Low"].str.replace('$', ''), errors='coerce'))) * 100
    df["52W % from High"] = df["52W % from High"].apply(lambda x: f"{x:.2f}%" if pd.notna(x) else None)
    return df

def sort_by_pe_ratio(df):
    """Sorts the DataFrame by P/E Ratio."""
    df["P/E Ratio"] = df["P/E Ratio"].astype(str).str.replace('$', '').str.replace('%', '')
    df["P/E Ratio"] = pd.to_numeric(df["P/E Ratio"], errors='coerce')
    return df.sort_values(by=["P/E Ratio"], ascending=True)

def display_stock_table(df):
    """Displays the stock data in a styled HTML table."""
    css_style = """
    <style>
    .table-bordered {
        width: auto;
        border-collapse: collapse;
        display: block;
        overflow-x: auto;
    }
    .table-bordered th, .table-bordered td {
        text-align: center !important;
        white-space: nowrap;
        padding: 8px;
        border: 1px solid #dee2e6;
    }
    .table-bordered th {
        font-weight: bold;
    }
    .positive {
        color: green;
    }
    .negative {
        color: red;
    }
    </style>
    """

    def color_change(val):
        try:
            val_num = float(val.replace('%', ''))
            if val_num > 0:
                return '<span class="positive">{}</span>'.format(val)
            elif val_num < 0:
                return '<span class="negative">{}</span>'.format(val)
            else:
                return val
        except ValueError:
            return val

    html = df.to_html(classes='table-bordered', formatters={'Day Change (%)': color_change}, escape=False)
    display(HTML(css_style + html))

def plot_pe_ratio_comparison(df):
    """Plots the P/E Ratio comparison of selected stocks."""
    plt.figure(figsize=(20, 5))
    width = 0.5
    pe_ratios_for_plot = pd.to_numeric(df["P/E Ratio"], errors='coerce')

    plt.bar(df["Company"], pe_ratios_for_plot, color='skyblue', width=width)
    plt.xlabel("Company")
    plt.ylabel("P/E Ratio")
    plt.title("P/E Ratio Comparison of Selected Stocks")
    plt.xticks(rotation=45, ha='right')
    plt.grid(axis="y", linestyle="--", alpha=0.7)
    plt.tight_layout()
    plt.show()

# Main Execution
if __name__ == "__main__":
    print_metric_definitions()
    stock_data = fetch_all_stock_data(stock_symbols)
    df = pd.DataFrame(stock_data)
    df = calculate_52w_percent_from_high(df)
    df_sorted = sort_by_pe_ratio(df)

    print("\n📊 Stock Comparison Based on Key Metrics:\n")
    display_stock_table(df_sorted)

    plot_pe_ratio_comparison(df_sorted)
```
